```python
from transformers import pipeline

# Load a pretrained text-generation model
generator = pipeline(
    "text-generation",
    model="openai-community/gpt2"
)

print("GPT-2 loaded successfully!")

```

```python
# Start with a product description prompt
prompt = "The Logitech G502 Wired Mouse let's gamers and other people in different professions do"

# Ask GPT-2 to continue the text
result = generator(
    prompt,
    max_new_tokens=120, # Changed from 60 to 120
    do_sample=True,
    temperature=0.7, # Changed from 0.8 to 0.7
    top_k=50,
    num_return_sequences=1,
    pad_token_id=generator.tokenizer.eos_token_id
)

print(result[0]["generated_text"])
```
